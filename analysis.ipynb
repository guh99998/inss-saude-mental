# Auxílios-doença e saúde mental

Os dados são gerados por `scripts/process_data.py`, que trata os arquivos do INSS (dez/2018 a mai/2023) e mantém só os auxílios-doença (espécies 31 e 91).

```python
import pandas as pd
```

## Carregando os dados

```python
colunas_categoricas = [
    'especie_beneficio', 'tipo_despacho', 'sexo', 'tipo_clientela', 'codigo_municipio_inss',
    'uf_residencia', 'nome_municipio', 'vinculo_dependente', 'forma_filiacao', 'uf'
]

df = pd.read_csv(
    'data/processado/auxilio_doenca_2018_2023.csv',
    parse_dates=['competencia_concessao', 'data_nascimento'],
    dtype={coluna: 'str' for coluna in colunas_categoricas} | {
        'cid': 'string',
        'descricao_cid': 'string',
        'quantidade_salarios_minimos_rmi': 'float64',
    },
)

# Convertidas depois da leitura: em blocos, o read_csv pode inferir tipos diferentes para as categorias
df[colunas_categoricas] = df[colunas_categoricas].astype('category')
```

## Conferência dos dados

```python
print("Registros:", len(df))
print("Memória (MB):", round(df.memory_usage(deep=True).sum() / 1024 ** 2))
print(df.dtypes)
```
